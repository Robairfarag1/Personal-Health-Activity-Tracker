```python
import matplotlib.pyplot as plt
import TimeSeriesHelperFunctions as tshf
import keras
import pandas as pd
import numpy as np

# TRAINING CONSTANTS:
NUM_EPOCHS = 20
BATCH_SIZE = 500
VALIDATION_SPLIT = 0.2

# KERAS RANDOM SEED FOR REPRODUCIBILITY
keras.utils.set_random_seed(1234)

# DATASET PATH 
ROOT_DIR = './Dataset'
```

```python
sensor_names = ['body_acc', 'body_gyro', 'total_acc']
loss_functions = ['mse', 'mase_loss_multi_step', 'trace_mse_loss']
metrics_xyz_combined = {}
 
for sensor_name in sensor_names:
    for loss_function in loss_functions:
        model, model_path, history, sensor_scaler, metrics_on_test = tshf.time_series_predictor_pipeline(
            root_dir = ROOT_DIR,
            sensor_name = sensor_name,
            user_loss_fn = loss_function,
            model_idx = 0,
            num_epochs= NUM_EPOCHS,
            batch_size=BATCH_SIZE,
            val_split=VALIDATION_SPLIT,
            debug_mode = False)
        
        metrics_xyz_combined[f'{sensor_name}_{loss_function}'] = metrics_on_test['xyz-Combined']
    

pd.DataFrame(metrics_xyz_combined)
```
